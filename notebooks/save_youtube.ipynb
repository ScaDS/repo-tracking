````python
%%bob
Load videos_with_authors.csv and convert the columns to resources in ../resources/scadsai.yml .
The CSV contains these columns:
* title (will be name in yml)
* description
* url
* upload_date (will be publication_date in yaml)
* url
* authors (" and "-separated)

Yml format:
```yml
resources:
- authors:
  - Firstname lastname
  - Firstname lastname
  description: '...'
  license: Unknown
  name: 'Smart Slide Generation: Re-Using Training Materials Through the Power of
    LLMs'
  num_downloads: 94
  publication_date: '2025-11-06'
  submission_date: <now>
  tags:
  - Living Lab Lecture
  - Video
  url:
  - https://doi.org/10.5281/zenodo.17541532
```

````

````python
import pandas as pd
import yaml
from datetime import date
````

````python
# 1. Load CSV
df = pd.read_csv("videos_with_authors.csv")
````

````python
# 2. Build resource entries
today_str = date.today().isoformat()

new_resources = []
for _, row in df.iterrows():
    authors = [a.strip() for a in str(row["authors"]).split(" and ")]
    resource = {
        "authors": authors,
        "description": str(row["description"]),
        "license": "Unknown",
        "name": str(row["title"]),
        "num_downloads": 0,
        "publication_date": str(row["upload_date"]),
        "submission_date": today_str,
        "tags": ["Living Lab Lecture", "Video"],
        "url": [str(row["url"])],
    }
    new_resources.append(resource)
````

````python
# 3. Load existing YAML
yml_path = "../resources/scadsai.yml"
try:
    with open(yml_path, "r", encoding="utf-8") as f:
        data = yaml.safe_load(f) or {}
except FileNotFoundError:
    data = {}

if "resources" not in data or data["resources"] is None:
    data["resources"] = []
````

````python
# 4. Append new resources
data["resources"].extend(new_resources)
````

````python
# 5. Save back to YAML
with open(yml_path, "w", encoding="utf-8") as f:
    yaml.dump(data, f, sort_keys=False, allow_unicode=True, default_flow_style=False)

print(f"Added {len(new_resources)} resources to {yml_path}")
````
